```python
nombre = input("Digite el nombre del estudiante: ")

calificacion1 = float(input("Digite la primera calificación: "))
calificacion2 = float(input("Digite la segunda calificación: "))
calificacion3 = float(input("Digite la tercera calificación: "))

promedio = (calificacion1 + calificacion2 + calificacion3) / 3

print(f"Estudiante: {nombre}")
print(f"Promedio: {promedio:.2f}")

if promedio >= 90:
    print("¡Excelente! Aprobó con honores (promedio mayor o igual a 90).")
elif promedio >= 70:
    print("Aprobó el curso.")
else:
    print("Reprobó el curso.")
```
