```python
#sklearn es una biblioteca de Python especializada en Machine Learning

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd

#Datos

X = [
    [18,1],
    [20,2],
    [25,5],
    [30,7],
    [35,9]
]


#Respuestas

y = [
    "No",
    "No",
    "Si",
    "Si",
    "Si"
]


#Crear el modelo
modelo = DecisionTreeClassifier()

#Entrenar fit() significa "Aprender con estos datos."

modelo.fit(X,y)

#Nuevo cliente: Aqui se crea un nuevo cliente

nuevo_cliente = [[27,6]]

#Prediccion con modelo.predict le pedimo a Python, que de una prediccion.

prediccion = modelo.predict(nuevo_cliente)

#Imprimimos la prediccion

print("Prediccion:",prediccion[0])


```

```python
#Esta libreria trae herraienta que sirve para dividir nuestros datos en dos grupos
from sklearn.model_selection import train_test_split
#Con este se importa el arbol de decision
from sklearn.tree import DecisionTreeClassifier
#Este sirve para medir la exactitud del modelo. ¿Cuantas veces acerto la computadora
from sklearn.metrics import accuracy_score

X = [
    [18, 1],
    [20, 2],
    [22, 3],
    [25, 5],
    [27, 6],
    [30, 7],
    [32, 8],
    [35, 9],
    [40, 10],
    [45, 12]
]

y = [
    "No",
    "No",
    "No",
    "Sí",
    "Sí",
    "Sí",
    "Sí",
    "Sí",
    "Sí",
    "Sí"
]

# Separar datos
#test_size=0.3 Significa: 30% de los datos serán para probar.
#random_state=42 sirve para controlar el componente aleatorio de la división.
#train_test_split mezcla los datos antes de separarlos.
#X_entrenamiento Datos que el modelo utilizará para aprender.
#y_entrenamiento Las respuestas correspondientes a esos datos.
#X_prueba Datos que el modelo no debe utilizar para aprender.
#y_prueba Las respuestas verdaderas que utilizaremos para comprobar si acertó.

X_entrenamiento, X_prueba, y_entrenamiento, y_prueba = train_test_split(
    X, y, test_size=0.9, random_state=17
)

#Crear el modelo
modelo = DecisionTreeClassifier()

#Entrenar fit() significa "Aprender con estos datos."

modelo.fit(X_entrenamiento,y_entrenamiento)

#Prediccion

predicciones = modelo.predict(X_prueba)

#Medir la excactitud

precision = accuracy_score(y_prueba,predicciones)

print("Predicciones:", predicciones)
print("Presupuesto reales:", y_prueba)
print("Precision:", precision)





```
