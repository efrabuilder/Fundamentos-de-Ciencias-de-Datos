```python
import pandas as pd
import matplotlib.pyplot as plt

# Datos en formato "largo" (una fila por mes y canal), como saldria de un dataset real
data = {
    "Mes": list(range(1, 13)) * 2,
    "Canal": ["Tienda_fisica"] * 12 + ["En_linea"] * 12,
    "Total": [4200, 4150, 4300, 4180, 4250, 4100, 4220, 4190, 4260, 4150, 4300, 4230] +
             [800, 870, 950, 1050, 1180, 1300, 1450, 1600, 1750, 1900, 2100, 2350],
}
df = pd.DataFrame(data)

ventas_mes = df.groupby(["Mes", "Canal"])["Total"].sum().unstack()
print("Ventas por mes y canal:")
print(ventas_mes)
print()

# Porcentaje en linea sobre el total
pct = (ventas_mes["En_linea"] / (ventas_mes["Tienda_fisica"] + ventas_mes["En_linea"])) * 100

print("Porcentaje de ventas en linea:")
print(pct.round(2))
print()
```

```python
# Igual que tu ejemplo: agrupar y sumar
ventas_mes = df.groupby(["Mes", "Canal"])["Total"].sum().unstack()

ventas_mes.plot(
    kind="line",
    figsize=(10, 5),
    marker="o",
    color=["#1f77b4", "#ff7f0e"]
)

plt.title("Ventas por Mes")
plt.xlabel("Mes")
plt.ylabel("Total de Ventas")
plt.xticks(rotation=-45)
plt.show()

# Porcentaje en linea sobre el total, con la misma logica de groupby
pct = (ventas_mes["En_linea"] / (ventas_mes["Tienda_fisica"] + ventas_mes["En_linea"])) * 100

pct.plot(
    kind="line",
    figsize=(10, 5),
    marker="o",
    color="green"
)

plt.title("Porcentaje de Ventas en Linea")
plt.xlabel("Mes")
plt.ylabel("Porcentaje (%)")
plt.xticks(rotation=-45)
plt.show()
```

```python
# Igual que tu ejemplo: agrupar y sumar
ventas_mes = df.groupby(["Mes", "Canal"])["Total"].sum().unstack()

# Porcentaje en linea sobre el total
pct = (ventas_mes["En_linea"] / (ventas_mes["Tienda_fisica"] + ventas_mes["En_linea"])) * 100

# Todo en un mismo grafico: ventas (eje izquierdo) + porcentaje (eje derecho)
fig, ax1 = plt.subplots(figsize=(11, 6))

ventas_mes.plot(
    kind="line", marker="o", ax=ax1,
    color=["#1f77b4", "#ff7f0e"]
)
ax1.set_ylabel("Total de Ventas")
ax1.set_xlabel("Mes")
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
pct.plot(
    kind="line", marker="s", linestyle="--", ax=ax2, color="green", label="Pct_en_linea"
)
ax2.set_ylabel("Porcentaje (%)")

# Etiquetas de valor en cada punto
for x, y in zip(ventas_mes.index, ventas_mes["Tienda_fisica"]):
    ax1.annotate(f'{y:,.0f}', (x, y), textcoords="offset points", xytext=(0, 8),
                 ha='center', fontsize=8, color='#1f77b4')

for x, y in zip(ventas_mes.index, ventas_mes["En_linea"]):
    ax1.annotate(f'{y:,.0f}', (x, y), textcoords="offset points", xytext=(0, -14),
                 ha='center', fontsize=8, color='#ff7f0e')

for x, y in zip(pct.index, pct):
    ax2.annotate(f'{y:.1f}%', (x, y), textcoords="offset points", xytext=(0, 8),
                 ha='center', fontsize=8, color='green')

# Leyenda unica, fuera del area de trazado
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
if ax1.get_legend(): ax1.get_legend().remove()
if ax2.get_legend(): ax2.get_legend().remove()

fig.legend(lines1 + lines2, labels1 + labels2,
           loc="upper center", bbox_to_anchor=(0.5, 1.08),
           ncol=3, frameon=False)

plt.title("Ventas por Mes y Porcentaje en Linea", pad=30)
plt.xticks(rotation=-45)
fig.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Datos
df = pd.DataFrame({
    "Mes": list(range(1, 13)),
    "Tienda_fisica": [4200, 4150, 4300, 4180, 4250, 4100, 4220, 4190, 4260, 4150, 4300, 4230],
    "En_linea": [800, 870, 950, 1050, 1180, 1300, 1450, 1600, 1750, 1900, 2100, 2350],
})

df["Pct_del_total"] = df["En_linea"] / (df["Tienda_fisica"] + df["En_linea"]) * 100
df["Pct_vs_fisica"] = df["En_linea"] / df["Tienda_fisica"] * 100

df = df.set_index("Mes")

fig, ax1 = plt.subplots(figsize=(12, 7))

df[["Tienda_fisica", "En_linea"]].plot(
    kind="line", marker="o", ax=ax1,
    color=["#1f77b4", "#ff7f0e"]
)
ax1.set_ylabel("Ventas (miles ₡)")
ax1.set_xlabel("Mes")
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
df[["Pct_del_total", "Pct_vs_fisica"]].plot(
    kind="line", marker="s", linestyle="--", ax=ax2,
    color=["green", "red"]
)
ax2.set_ylabel("Porcentaje (%)")

# Etiquetas de valor en cada punto
for x, y in zip(df.index, df["Tienda_fisica"]):
    ax1.annotate(f'{y:,}', (x, y), textcoords="offset points", xytext=(0, 8),
                 ha='center', fontsize=8, color='#1f77b4')

for x, y in zip(df.index, df["En_linea"]):
    ax1.annotate(f'{y:,}', (x, y), textcoords="offset points", xytext=(0, -14),
                 ha='center', fontsize=8, color='#ff7f0e')

for x, y in zip(df.index, df["Pct_del_total"]):
    ax2.annotate(f'{y:.1f}%', (x, y), textcoords="offset points", xytext=(0, 8),
                 ha='center', fontsize=7, color='green')

for x, y in zip(df.index, df["Pct_vs_fisica"]):
    ax2.annotate(f'{y:.1f}%', (x, y), textcoords="offset points", xytext=(0, -14),
                 ha='center', fontsize=7, color='red')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
if ax1.get_legend(): ax1.get_legend().remove()
if ax2.get_legend(): ax2.get_legend().remove()

fig.legend(lines1 + lines2, labels1 + labels2,
           loc="upper center", bbox_to_anchor=(0.5, 1.08),
           ncol=4, frameon=False)

plt.title("Ventas por Mes y Porcentajes en Linea", pad=30)
plt.xticks(rotation=-45)
fig.tight_layout()
plt.show()
```

```python
# Analisis y conclusiones (celda autocontenida)
import pandas as pd

# Reconstruir el DataFrame por si esta celda se corre sola
df = pd.DataFrame({
    "Mes": list(range(1, 13)),
    "Tienda_fisica": [4200, 4150, 4300, 4180, 4250, 4100, 4220, 4190, 4260, 4150, 4300, 4230],
    "En_linea": [800, 870, 950, 1050, 1180, 1300, 1450, 1600, 1750, 1900, 2100, 2350],
})
df["Pct_del_total"] = df["En_linea"] / (df["Tienda_fisica"] + df["En_linea"]) * 100
df["Pct_vs_fisica"] = df["En_linea"] / df["Tienda_fisica"] * 100
df = df.set_index("Mes")

crecimiento_en_linea = (df["En_linea"].iloc[-1] / df["En_linea"].iloc[0] - 1) * 100
crecimiento_fisica = (df["Tienda_fisica"].iloc[-1] / df["Tienda_fisica"].iloc[0] - 1) * 100
variacion_fisica = df["Tienda_fisica"].max() - df["Tienda_fisica"].min()

pct_inicial = df["Pct_del_total"].iloc[0]
pct_final = df["Pct_del_total"].iloc[-1]

mes_max_en_linea = df["En_linea"].idxmax()

print("=" * 60)
print("CONCLUSIONES DEL ANÁLISIS DE VENTAS: TIENDA FÍSICA VS EN LÍNEA")
print("=" * 60)
print(f"- La Tienda Física se mantiene prácticamente estable durante el año, "
      f"oscilando entre {df['Tienda_fisica'].min():,} y {df['Tienda_fisica'].max():,} "
      f"(variación de solo {variacion_fisica:,}), lo que indica un canal maduro "
      f"sin crecimiento ni caída relevante ({crecimiento_fisica:+.1f}% de enero a diciembre).")
print(f"- Las ventas En Línea crecen de forma sostenida mes a mes, pasando de "
      f"{df['En_linea'].iloc[0]:,} en el mes 1 a {df['En_linea'].iloc[-1]:,} en el mes 12, "
      f"un incremento acumulado de {crecimiento_en_linea:.0f}%.")
print(f"- El peso de En Línea sobre el total de ventas pasó de {pct_inicial:.1f}% "
      f"a {pct_final:.1f}% en el año, mostrando una ganancia de participación "
      f"de {pct_final - pct_inicial:.1f} puntos porcentuales.")
print(f"- El mes con mayor volumen de ventas en línea fue el mes {mes_max_en_linea} "
      f"({df['En_linea'].loc[mes_max_en_linea]:,}), coincidiendo con el cierre de año.")
print("- Conclusión: la tendencia es clara y sostenida, no estacional ni aislada: "
      "el canal En Línea está ganando terreno de forma constante mes tras mes, "
      "mientras la Tienda Física se mantiene sin cambios estructurales. "
      "De continuar este ritmo, el canal en línea podría superar en participación "
      "relativa a la tienda física en el mediano plazo.")
print("=" * 60)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Creacion de datos
data = {
    "Trimestre": ["Q1 (ene-mar)", "Q2 (abr-jun)", "Q3 (jul-sep)", "Q4 (oct-dic)"] * 3,   #Diccionario
    "Anio": ["Año 1"] * 4 + ["Año 2"] * 4 + ["Año 3"] * 4,
    "Renuncias": [3, 2, 2, 9,
                  4, 3, 2, 10,
                  3, 2, 3, 11]
}
df = pd.DataFrame(data)

# Tabla pivote: Trimestre en filas, Año en columnas
renuncias_trimestre = df.pivot(index="Trimestre", columns="Anio", values="Renuncias")
renuncias_trimestre = renuncias_trimestre.reindex(
    ["Q1 (ene-mar)", "Q2 (abr-jun)", "Q3 (jul-sep)", "Q4 (oct-dic)"]
)

#Creacion de graficos

ax = renuncias_trimestre.plot(
    kind="line",   #Grafico de lineas         #Grafico que compara renuncias por trimestre entre años
    figsize=(10, 5),
    marker="o",
    linewidth=2
)

plt.title("Renuncias por Trimestre - Equipo de Bodega (líneas por año)")
plt.xlabel("Trimestre")
plt.ylabel("Cantidad de Renuncias")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

# Etiquetas con el valor de cada punto (bolita)
for columna in renuncias_trimestre.columns:
    for i, valor in enumerate(renuncias_trimestre[columna]):
        ax.annotate(str(valor), (i, valor), textcoords="offset points",
                    xytext=(0, 8), ha="center", fontsize=9)

plt.show()
```

```python
#Creacion de graficos

ax2 = renuncias_trimestre.plot(
    kind="bar",   #Grafico de Barras         #Grafico que compara renuncias por trimestre agrupadas
    figsize=(10, 5),
    color=["#3B82F6", "#22C55E", "#F97316"]
)

plt.title("Renuncias por Trimestre - Equipo de Bodega (agrupadas)")
plt.xlabel("Trimestre")
plt.ylabel("Cantidad de Renuncias")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

# Etiquetas con el valor de cada barra
for contenedor in ax2.containers:
    ax2.bar_label(contenedor, fontsize=9)

plt.show()
```

```python
# Porcentaje de renuncias por trimestre sobre el total de cada año

renuncias_pct = renuncias_trimestre.div(renuncias_trimestre.sum(axis=0), axis=1) * 100
renuncias_pct = renuncias_pct.round(1)

print("Porcentaje de renuncias por trimestre (sobre el total de cada año):")
print(renuncias_pct)
```

```python
# Analisis numerico

q1_q3 = df[df["Trimestre"] != "Q4 (oct-dic)"]["Renuncias"]
q4 = df[df["Trimestre"] == "Q4 (oct-dic)"]["Renuncias"]

promedio_q1_q3 = q1_q3.mean()
promedio_q4 = q4.mean()
incremento = (promedio_q4 / promedio_q1_q3 - 1) * 100

print("=" * 60)
print("CONCLUSIONES DEL ANÁLISIS DE ROTACIÓN - EQUIPO DE BODEGA")
print("=" * 60)
print(f"- Q1, Q2 y Q3 se mantienen bajos y estables en los 3 años, "
      f"entre {q1_q3.min()} y {q1_q3.max()} renuncias por trimestre "
      f"(promedio: {promedio_q1_q3:.1f}).")
print(f"- Q4 se dispara de forma consistente cada año: "
      f"{', '.join(map(str, q4))} renuncias (promedio: {promedio_q4:.1f}), "
      f"muy por encima del resto de trimestres.")
print(f"- En Q4 las renuncias son en promedio un {incremento:.0f}% más altas "
      f"que en el resto del año.")
print("- El pico ocurre en la MISMA época (oct-dic) los tres años consecutivos, "
      "no en trimestres distintos ni de forma aleatoria.")
print("- Conclusión: el patrón corresponde a ESTACIONALIDAD, no a un evento "
      "aislado ni a una tendencia general de crecimiento de renuncias en la empresa. "
      "Las causas probables deben buscarse en factores propios de fin de año "
      "(ej. temporada alta/carga operativa, aguinaldo y cierre de año, bonos, "
      "búsqueda de nuevo empleo antes de fin de año), y no en una falla estructural "
      "constante durante todo el período.")
print("=" * 60)
```
