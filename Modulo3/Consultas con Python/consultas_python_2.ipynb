```python
#Calcular el total de ventas

df["total_ventas"] = df["Cantidad"] * df["Precio_Unitario"]
df[["ID_Venta", "Producto", "Cantidad", "Precio_Unitario", "Total"]]
```

```python
#Obtener el total vendido por producto

df["Total"] = df["Cantidad"] * df["Precio_Unitario"]
df.groupby("Producto")["Total"].sum()
df.groupby("Producto", as_index=False)["Total"].sum()
```

```python
#Obtener el total vendido por metodo de pago

df["Total"] = df["Cantidad"] * df["Precio_Unitario"]
df.groupby("Metodo_Pago")["Total"].sum()
df.groupby("Metodo_Pago", as_index=False)["Total"].sum()
```

```python
#Conteo de frecuencias
df["Ciudad"].value_counts()
df["Producto"].value_counts()
df["Categoria"].value_counts()
df["Metodo_Pago"].value_counts()
df["Vendedor"].value_counts()
```

```python
#El que más aparece / el que menos aparece
df["Ciudad"].value_counts().idxmax()
df["Ciudad"].value_counts().idxmin()
df["Producto"].value_counts().head(1)
df["Producto"].value_counts().tail(1)
df["Categoria"].mode()[0]
```

```python
#Cantidad total, promedio, conteo, min y max por ciudad
df.groupby("Ciudad")["Cantidad"].sum()
df.groupby("Ciudad")["Cantidad"].mean()
df.groupby("Ciudad")["Cantidad"].count()
df.groupby("Ciudad")["Cantidad"].min()
df.groupby("Ciudad")["Cantidad"].max()
df.groupby("Ciudad")["ID_Venta"].count()
```

```python
#Columna nueva: Total_Venta
df["Total_Venta"] = df["Cantidad"] * df["Precio_Unitario"]
```

```python
#groupby() con agg() — la que "hace todo" en una sola consulta

df.groupby("Ciudad").agg({
    "Cantidad": ["sum", "mean", "count", "min", "max"],
    "Precio_Unitario": ["sum", "mean", "count", "min", "max"],
    "Total_Venta": ["sum", "mean"]
})
```

```python
#groupby() con dos niveles
df.groupby(["Ciudad", "Categoria"])["Cantidad"].sum()
df.groupby(["Ciudad", "Categoria"])["Cantidad"].agg(["sum", "mean", "count", "min", "max"])
df.groupby(["Ciudad", "Producto"])["Total_Venta"].sum()
df.groupby(["Vendedor", "Metodo_Pago"])["Total_Venta"].sum()
```

```python
#pivot_table() — la consulta final, la más "profesional"
pd.pivot_table(df, index="Ciudad", columns="Categoria", values="Cantidad", aggfunc="sum")
pd.pivot_table(df, index="Ciudad", columns="Categoria", values="Total_Venta", aggfunc="sum")
pd.pivot_table(df, index="Vendedor", columns="Metodo_Pago", values="Total_Venta", aggfunc="sum")
```
